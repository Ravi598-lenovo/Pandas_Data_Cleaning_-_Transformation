```python
# ----------------------------------------------------------
# 1. Import Required Libraries
# ----------------------------------------------------------

import pandas as pd
import numpy as np
import seaborn as sns

# Set diaplay options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 100)
```

```python
# ----------------------------------------------------------
# 2. Load Dataset from Library
# ----------------------------------------------------------

# Load Dataset directly from seaborn
df = sns.load_dataset('titanic')

# Display first 5 rows
df.head()
```

```python
# ----------------------------------------------------------
# 3. Understand the Dataset Structure
# ----------------------------------------------------------

# Shape of the dataset (rows, columns)
df.shape

# Column names
df.columns

# Data types and non-null counts
df.info()
```

```python
# ----------------------------------------------------------
# 4. Identify Missing Values
# ----------------------------------------------------------

# Check how many missing values are present in each column
df.isnull().sum()
```

```python
# ----------------------------------------------------------
# 5. Handling Missing Values
# ----------------------------------------------------------

# (First) Fill missing 'age' values with median
df['age'] = df['age'].fillna(df['age'].median())

# (Second) Fill missing 'embarked' values with mode
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# (Third) Drop 'deck' column (Why? because it has too many missing values)
df = df.drop(columns=['deck'])

# Now verify missing values again
df.isnull().sum()
```

```python
# ----------------------------------------------------------
# 6. Remove Duplicate Rows (if any)
# ----------------------------------------------------------

# Check duplicate rows
df.duplicated().sum()

# Drop duplicates
df = df.drop_duplicates()
```

```python
# ----------------------------------------------------------
# 7. Data Type Conversion
# ----------------------------------------------------------

# Convert categorical columns to 'category' type
categorical_cols = ['sex', 'class', 'embarked', 'who', 'alive']

for col in categorical_cols:
    df[col] = df[col].astype('category')

# Check updated data types
df.info()
```

```python
# ----------------------------------------------------------
# 8. Feature Engineering (Create New Features)
# ----------------------------------------------------------

# (First) Create 'family-size' feature
df['family_size'] = df['sibsp'] + df['parch'] + 1

# (Second) Create 'is_alone' feature
df['is_alone'] = np.where(df['family_size'] == 1, 1, 0)

# (Third) Create age group feature
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 12, 20, 40, 60, 80],
    labels=['Child', 'Teen', 'Adult', 'Middle-Aged', 'Senior']
)

df.head()
```

```python
# ----------------------------------------------------------
# 9. String Cleaning & Standardization
# ----------------------------------------------------------

# Convert 'embarked' values to lowercase
df['embarked'] = df['embarked'].str.lower()

# Rename column names to snake_case
df.columns = df.columns.str.lower().str.replace(' ', '_')

df.head()
```

```python
# ----------------------------------------------------------
# 10. Outlier Detection & Handling
# ----------------------------------------------------------

# Detect outliers in 'fare' using IQR method
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1

# Define upper limit
upper_bound = Q3 + 1.5 * IQR

# Cap extreme values
df['fare'] = np.where(df['fare'] > upper_bound, upper_bound, df['fare'])

# Verify outliers
df['fare'].describe()
```

```python
# ----------------------------------------------------------
# 11. Advanced Group Operations
# ----------------------------------------------------------

# Average fare by passenger class
df.groupby('class')['fare'].mean()

# Survival rate by gender
df.groupby('sex')['survived'].mean()

# Average fare by age group
df.groupby('age_group')['fare'].mean()
```

```python
# ----------------------------------------------------------
# 12. Final Clean Dataset Overview
# ----------------------------------------------------------

df.describe(include='all')
```
